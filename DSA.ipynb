```python
n = 5

# Upper Half
for i in range(1, n + 1):

    # Left Hollow Triangle
    for j in range(i):
        if j == 0 or j == i - 1 or i == n:
            print("*", end="")
        else:
            print(" ", end="")

    # Middle Spaces
    for j in range(2 * (n - i)):
        print(" ", end="")

    # Right Hollow Triangle
    for j in range(i):
        if j == 0 or j == i - 1 or i == n:
            print("*", end="")
        else:
            print(" ", end="")

    print()


# Lower Half
for i in range(1, n):

    # Left Hollow Inverted Triangle
    for j in range(n - i):
        if j == 0 or j == (n - i) - 1:
            print("*", end="")
        else:
            print(" ", end="")

    # Middle Spaces
    for j in range(2 * i):
        print(" ", end="")

    # Right Hollow Inverted Triangle
    for j in range(n - i):
        if j == 0 or j == (n - i) - 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()
```

```python
#Linear Search
arr = [10,20,30,40,50]
key = 200
found = False
for i in range(len(arr)):
  if arr[i] == key:
    print(f"Element found at index: {i}")
    found = True
if found == False:
    print("Element not found")
```

```python
#Max element in array
arr = [10,12,15,18,20]
max_num = arr[0]
for i in range(len(arr)):
  if arr[i] > max_num:
    max_num = arr[i]
print(f"Maximum element in array: {max_num}")
```

```python
#Two pointers problem or reversing array
arr = [10,20,30,40,50,60,70]
left = 0
right = len(arr) - 1
while left < right:
  arr[left], arr[right] = arr[right], arr[left]
  left += 1
  right -= 1
print(arr)
```

```python
from typing_extensions import is_protocol
#Checking array is sorted or not
arr = [10,20,35,40,50]
is_sorted = True
for i in range(len(arr)-1):
    if arr[i] > arr[i+1]:
      is_sorted = False
      break
if is_sorted:
  print("Array is sorted")
else:
  print("Array is not sorted")
```

```python
#Counting even and odd
arr = [10, 13, 24, 7, 9, 8]
eve = 0
odd = 0
for i in range(len(arr)):
  if arr[i] % 2 == 0:
    eve += 1
  else:
    odd += 1
print(f"Even numbers are: {eve}")
print(f"Odd numbers are: {odd}")
```

```python
#Finding the second largest of array
arr = [10,20,30,40,50]
largest = arr[0]
second_largest = arr[0]
for i in range(len(arr)):
  if largest < arr[i]:
    second_largest = largest
    largest = arr[i]
  elif second_largest < arr[i] and arr[i] != largest:
    second_largest = arr[i]
print(f"Second largest element in array: {second_largest}")
```

```python
#Moving zeroes to end
arr = [0, 1, 0, 3, 12]
left = 0
for i in range(len(arr)):
  if arr[i] != 0:
    arr[left], arr[i] = arr[i], arr[left]
    left += 1
print(arr)
```

```python
# Rotate array to the right by one position

arr = [10, 20, 30, 40, 50]

temp = arr[-1]

for i in range(len(arr) - 1, 0, -1):
    arr[i] = arr[i - 1]

arr[0] = temp

print(arr)
```

```python
#Removing duplicates
arr = [1,1,2,2,3,3,4,5,5]
left = 0
for i in range(1,len(arr)):
  if arr[left] != arr[i]:
    arr[left + 1] = arr[i]
    left += 1
print(arr[:left+1])
```

```python
#Missing number
arr = [1,2,4,5]
actual_sum = 0
n = len(arr) + 1
for i in range(len(arr)):
  actual_sum += arr[i]
expected_sum = (n*(n+1)) // 2
missing_number = expected_sum - actual_sum
print(missing_number)
```

```python
#Frequency count
arr = [1,2,2,3,1,2]
frequency = {}
for i in range(len(arr)):
  num = arr[i]
  if num in frequency:
    frequency[num] += 1
  else:
      frequency[num] = 1
print(frequency)

max_frequency = 0
most_frequency = 0
for num in frequency:
    if frequency[num] > max_frequency:
        max_frequency = frequency[num]
        most_frequent = num

print("Most Frequent Element:", most_frequent)
print("Frequency:", max_frequency)
```
